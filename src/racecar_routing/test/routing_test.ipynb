```python
import rclpy
from rclpy.node import Node
from geometry_msgs.msg import PoseStamped
from nav_msgs.msg import Odometry
import time
from nav_msgs.msg import Path as routePath


#do ros2 topic echo /Routing/Path, and run next cell to receive message, therefore it should work and send a path
```

```python
try:
    rclpy.init()

    node_sender = Node('fake_publisher')

    odom_pub = node_sender.create_publisher(Odometry, '/slam_pose', 10)
    goal_pub = node_sender.create_publisher(PoseStamped, '/move_base_simple/goal', 10)

    odom_msg = Odometry()
    odom_msg.pose.pose.position.x = 2.0
    odom_msg.pose.pose.position.y = 0.15
    
    odom_msg.pose.pose.orientation.x = 0.0
    odom_msg.pose.pose.orientation.y = 0.0
    odom_msg.pose.pose.orientation.z = 0.0
    odom_msg.pose.pose.orientation.w = 1.0
    

    goal_msg = PoseStamped()
    goal_msg.header.frame_id = 'map'
    goal_msg.pose.position.x = 0.8
    goal_msg.pose.position.y = 5.0

    goal_msg.pose.orientation.x = 0.0
    goal_msg.pose.orientation.y = 0.0
    goal_msg.pose.orientation.z = 0.0
    goal_msg.pose.orientation.w = 1.0

    # Publish odom first, then goal
    odom_pub.publish(odom_msg)
    time.sleep(0.2)
    goal_pub.publish(goal_msg)

    node_sender.get_logger().info('Published fake odom and goal.')
finally:
    node_sender.destroy_node()
    rclpy.shutdown()




```
